```python
from langchain_text_splitters import RecursiveCharacterTextSplitter
```

```python
text = """A science-backed morning routine leverages your biological clock to maximize energy, mood, and focus for the rest of the day. Upon waking, immediate hydration reverses overnight fluid loss, while early exposure to natural sunlight triggers a critical cortisol spike that boosts alertness and resets your circadian rhythm.

Waiting 90 to 120 minutes before consuming caffeine allows your body to naturally clear adenosine, preventing the common afternoon crash. Incorporating brief physical movement increases core body temperature and releases dopamine, which sharpens mental clarity. Finally, protecting the first hour of the day from digital distractions allows you to utilize peak cognitive function for deep, impactful work"""

splitter = RecursiveCharacterTextSplitter(
    chunk_size=500,
    chunk_overlap=0
)
```

```python
chunks = splitter.split_text(text)

print(len(chunks))

print(chunks[0])                                 #1st chunk
print(chunks[1])                                 #2nd chunk
```
